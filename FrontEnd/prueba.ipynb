```python
import requests

base_url = "http://127.0.0.1:8000"

login_data = {
    "username": "prueba",
    "password": "prueba"
}

login_url = f"{base_url}/login"
login_response = requests.post(login_url, data=login_data, headers={"Content-Type": "application/x-www-form-urlencoded"})

if login_response.status_code == 200:
    print("Login exitoso")
    
    token = login_response.json().get("access_token")
    if token:
        print("Token recibido:", token)

        user_info_url = f"{base_url}/users/me"
        headers = {
            "Authorization": f"Bearer {token}",
        }
        user_info_response = requests.get(user_info_url, headers=headers)

        if user_info_response.status_code == 200:
            print("Información del usuario:")
            print(user_info_response.json())
        else:
            print(f"Error al obtener la información del usuario: {user_info_response.status_code}")
            print(user_info_response.json())
    else:
        print("No se recibió el token del login.")
else:
    print(f"Error al hacer login: {login_response.status_code}")
    print(login_response.json())
```
